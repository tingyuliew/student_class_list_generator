```python
import streamlit as st
from docx import Document
from copy import deepcopy
import random
import math
import tempfile
import os
```

```python
doc = Document('A391-002.docx')
```

```python
if len(doc.tables) == 0:
    print("No tables found in document.")
```

```python
source_table = doc.tables[0]

students = []

for row in source_table.rows:
    for cell in row.cells:

        text = cell.text.strip()

        if text:
            students.append(
                deepcopy(cell._tc)
            )

total_students = len(students)
```

```python
random.shuffle(students)
```

```python
group_size = 5
num_groups = math.ceil(
        total_students / group_size  #get the number of groups based on 5 students per group
)

groups = []

for i in range(num_groups):

    start = i * group_size
    end = start + group_size

    groups.append(
        students[start:end]
    )

output_doc = Document()

original_cols = len(source_table.columns)
```

```python

        for idx, group in enumerate(groups, start=1):

            output_doc.add_heading(
                f"Group {idx}",
                level=1
            )

            rows_needed = math.ceil(
                len(group) / original_cols
            )

            table = output_doc.add_table(
                rows=rows_needed,
                cols=original_cols
            )

            student_idx = 0

            for r in range(rows_needed):

                for c in range(original_cols):

                    if student_idx >= len(group):
                        continue

                    target_cell = table.cell(r, c)

                    target_cell._tc.clear_content()

                    copied_cell = deepcopy(
                        group[student_idx]
                    )

                    target_cell._tc.append(
                        copied_cell
                    )

                    student_idx += 1

            if idx != len(groups):
                output_doc.add_page_break()
```

```python
output_path = "grouped_students.docx"

output_doc.save(output_path)
```
